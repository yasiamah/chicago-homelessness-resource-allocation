# Chicago Homelessness Data Source Inventory

## Purpose

This notebook creates a structured inventory of the datasets planned for the Chicago Housing Instability and Resource Allocation project.

The inventory records what each source measures, its time period, how it will be used, and its main limitations.

## Measurement Rule

These datasets do not all measure the same thing. PIT measures literal homelessness on one night. ACS and CHAS measure household and housing conditions. The CCH estimate uses a broader annual definition that includes doubled-up homelessness.

```python
sources = pd.DataFrame(source_records)

print("Number of sources:", len(sources))
display(sources)
```

```python
expected_columns = [
    "source_name",
    "organization",
    "data_type",
    "coverage",
    "latest_period",
    "priority",
    "planned_use",
    "main_limitation",
    "access_url",
]
assert len(sources) == 7, "The inventory should contain seven sources."
assert list(sources.columns) == expected_columns, "The columns are incorrect or out of order."
assert sources["source_name"].is_unique, "The source_name column should have unique values."
assert not sources["access_url"].isnull().any(), "The access_url column should not contain null values."
assert sources["access_url"].str.startswith("https://").all(), "At least one URL is not secure."
print("All inventory validation checks passed.")
```

```python
print("Missing values by column:")
display(sources.isna().sum())

print("Duplicate source names:", sources["source_name"].duplicated().sum())
print("All links use HTTPS:", sources["access_url"].str.startswith("https://").all())
```
